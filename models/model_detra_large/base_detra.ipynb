```python
import json
from pathlib import Path

import numpy as np
from pycocotools.coco import COCO
from pycocotools.cocoeval import COCOeval
from tqdm.auto import tqdm

DATASET_DIR = Path("/kaggle/input/datasets/arinaparashchina/bdd-any-images/10k")

VAL_DIR = DATASET_DIR / "valid"
ANNOTATIONS = VAL_DIR / "_annotations.coco.json"

CLASS_MAPPING = {
    "bike": "bicycle",
    "bus": "bus",
    "car": "car",
    "motor": "motorcycle",
    "person": "person",
    "traffic light": "traffic light",
    "train": "train",
    "truck": "truck",
}


coco = COCO(str(ANNOTATIONS))

categories = {cat["id"]: cat["name"] for cat in coco.loadCats(coco.getCatIds())}


target_categories = {
    cat_id: CLASS_MAPPING[name]
    for cat_id, name in categories.items()
    if name in CLASS_MAPPING
}

print("Classes used for evaluation:")
for cat_id, name in target_categories.items():
    print(f"  {cat_id}: {categories[cat_id]} -> {name}")

print()


# ============================================================
# IMAGE LIST
# ============================================================

image_ids = coco.getImgIds()
images = coco.loadImgs(image_ids)

print("Images:", len(images))


# ============================================================
# RUN RF-DETR
# ============================================================

results = []

for img_info in tqdm(images):
    image_id = img_info["id"]

    image_path = VAL_DIR / img_info["file_name"]

    detections = model.predict(str(image_path))

    for xyxy, confidence, _class_id, class_name in zip(
        detections.xyxy,
        detections.confidence,
        detections.class_id,
        detections.data["class_name"],
        strict=False,
    ):
        class_name = str(class_name)

        # Оставляем только пересечение с твоими классами
        if class_name not in CLASS_MAPPING.values():
            continue

        # Находим ТВОЙ category_id по имени COCO-класса
        matching = [
            cat_id
            for cat_id, coco_name in target_categories.items()
            if coco_name == class_name
        ]

        if not matching:
            continue

        category_id = matching[0]

        x1, y1, x2, y2 = map(float, xyxy)

        # xyxy -> COCO xywh
        x = x1
        y = y1
        w = x2 - x1
        h = y2 - y1

        results.append(
            {
                "image_id": image_id,
                "category_id": category_id,
                "bbox": [x, y, w, h],
                "score": float(confidence),
            }
        )


print()
print("Predictions:", len(results))
```

```python
# ============================================================
# COCO EVALUATION
# ============================================================

coco_dt = coco.loadRes(results)

coco_eval = COCOeval(coco, coco_dt, "bbox")

# Оцениваем только наши 8 классов
target_cat_ids = list(target_categories.keys())

coco_eval.params.catIds = target_cat_ids

coco_eval.evaluate()
coco_eval.accumulate()
coco_eval.summarize()
```

```python
# ============================================================
# AP PER CLASS
# ============================================================

precision = coco_eval.eval["precision"]

# precision shape:
# [IoU, Recall, Category, Area, MaxDets]

print("\nAP per class:")

for idx, cat_id in enumerate(target_cat_ids):
    # IoU 0.50:0.95
    values = precision[:, :, idx, 0, 2]

    values = values[values > -1]

    if len(values) == 0:
        ap = float("nan")
    else:
        ap = np.mean(values)

    print(f"{categories[cat_id]:20s} AP={ap:.4f}")
```

```python
import cv2
import matplotlib.pyplot as plt

# Берём картинку
image_info = images[30]
image_path = VAL_DIR / image_info["file_name"]

# Предсказание
detections = model.predict(str(image_path))

# Читаем изображение

image = cv2.imread(str(image_path))
image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

# Рисуем боксы
for box, confidence, class_name in zip(
    detections.xyxy,
    detections.confidence,
    detections.data["class_name"],
    strict=False,
):
    x1, y1, x2, y2 = map(int, box)

    # bbox
    cv2.rectangle(
        image,
        (x1, y1),
        (x2, y2),
        (0, 255, 0),
        1,
    )

    # текст
    label = f"{class_name} {confidence:.2f}"

    cv2.putText(
        image,
        label,
        (x1, max(y1 - 5, 15)),
        cv2.FONT_HERSHEY_SIMPLEX,
        0.6,
        (0, 255, 0),
        1,
    )

# Показываем
plt.figure(figsize=(16, 9))
plt.imshow(image)
plt.axis("off")
plt.show()
```

```python
import cv2
import matplotlib.pyplot as plt
from pycocotools.coco import COCO

# ---------------------------------------------------------
# 1. Берём изображение
# ---------------------------------------------------------

image_info = images[1]

image_path = VAL_DIR / image_info["file_name"]

# ---------------------------------------------------------
# 2. Предсказания RF-DETR
# ---------------------------------------------------------

detections = model.predict(str(image_path))

# ---------------------------------------------------------
# 3. Загружаем изображение
# ---------------------------------------------------------

image = cv2.imread(str(image_path))
image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

# ---------------------------------------------------------
# 4. Исходная разметка COCO
# ---------------------------------------------------------

ann_ids = coco.getAnnIds(imgIds=[image_info["id"]], iscrowd=None)

annotations = coco.loadAnns(ann_ids)

categories = {cat["id"]: cat["name"] for cat in coco.loadCats(coco.getCatIds())}

# ---------------------------------------------------------
# 5. Рисуем GT
#    Красный = ground truth
# ---------------------------------------------------------

for ann in annotations:
    x, y, w, h = ann["bbox"]

    x1 = int(x)
    y1 = int(y)
    x2 = int(x + w)
    y2 = int(y + h)

    class_name = categories[ann["category_id"]]

    cv2.rectangle(
        image,
        (x1, y1),
        (x2, y2),
        (255, 0, 0),  # RED
        2,
    )

    cv2.putText(
        image,
        f"GT: {class_name}",
        (x1, max(y1 - 5, 15)),
        cv2.FONT_HERSHEY_SIMPLEX,
        0.55,
        (255, 0, 0),
        2,
    )

# ---------------------------------------------------------
# 6. Рисуем predictions
#    Зелёный = prediction
# ---------------------------------------------------------

for box, confidence, class_name in zip(
    detections.xyxy,
    detections.confidence,
    detections.data["class_name"],
    strict=False,
):
    x1, y1, x2, y2 = map(int, box)

    cv2.rectangle(
        image,
        (x1, y1),
        (x2, y2),
        (0, 255, 0),  # GREEN
        2,
    )

    label = f"Pred: {class_name} {confidence:.2f}"

    cv2.putText(
        image,
        label,
        (x1, min(y2 + 18, image.shape[0] - 5)),
        cv2.FONT_HERSHEY_SIMPLEX,
        0.55,
        (0, 255, 0),
        2,
    )

# ---------------------------------------------------------
# 7. Показываем
# ---------------------------------------------------------

plt.figure(figsize=(18, 10))
plt.imshow(image)
plt.title(f"{image_info['file_name']} | RED = GT, GREEN = prediction")
plt.axis("off")
plt.show()
```
